```python
#Part 1-Google Colab and Python
# 1. Create a numpy array of size 2 x 2
import numpy as np

array_2x2 = np.array([[1,2],[3,4]])

print("This is my numpy 2x2 array")
print(array_2x2)
```

```python
# 2. Create a random numpy array of size 3 x 4
#The first number is the row and second is the column
#So this is a grid of 3 rows and 4 columns
#this allows you to make an array of shape 3,4 and fill it with random numbers. Your numbers will be float
random3x4 = np.random.rand(3,4)

print("Random 3x4 array")
print(random3x4)
```

```python
# 3. Create a numpy array of ones and a numpy array of zeroes
ones_array = np.ones((3, 3))
zeros_array = np.zeros((3, 3))

print("Array of Ones:")
print(ones_array)

print("Array of zeros:")
print(zeros_array)
```

```python
# 4. Create a random numpy array of size 6x3 and transpose it
random6x3 = np.random.rand(6, 3)
transposed = random6x3.T

print("Random 6x3 array:")
print(random6x3)
print("Shape:", random6x3.shape)

print("Transposed (3x6) array:")
print(transposed)
print("Shape:", transposed.shape)
```

```python
# 5. Create two random numpy arrays of size n x p and p x m and compute the dot product
n, p, m = 2, 3, 4
A = np.random.rand(n, p)
B = np.random.rand(p, m)
dot_product = np.dot(A, B)

print("Array A (n x p):", A.shape)
print(A)
print("Array B (p x m):", B.shape)
print(B)
print("Dot product A . B (n x m):", dot_product.shape)
print(dot_product)
```
